```python
def count_total_dicts_in_folder(parent_folder):
    """Counts the total number of dictionaries in all 'videos_info.json' files inside the given folder and its subfolders.
       Also counts the number of child folders."""
    total_dicts = 0
    child_folders = 0

    # Ensure the parent folder exists
    if not os.path.exists(parent_folder):
        print(f"Error: Folder '{parent_folder}' does not exist.")
        return 0, 0

    # Walk through all subdirectories
    for root, dirs, files in os.walk(parent_folder):
        if root == parent_folder:
            child_folders = len(dirs)  # Count number of direct subfolders

        if "videos_info.json" in files:  # Only process 'videos_info.json' files
            file_path = os.path.join(root, "videos_info.json")

            try:
                with open(file_path, "r", encoding="utf-8") as file:
                    data = json.load(file)

                    # Ensure the content is a list of dictionaries
                    if isinstance(data, list) and all(isinstance(item, dict) for item in data):
                        count = len(data)
                        total_dicts += count
                        print(f"{file_path}: {count} dictionaries found.")
                    else:
                        print(
                            f"Skipping {file_path}: Not a list of dictionaries.")

            except (json.JSONDecodeError, ValueError):
                print(f"Skipping {file_path}: Invalid JSON format.")
            except Exception as e:
                print(f"Error reading {file_path}: {e}")

    print(
        f"\nTotal dictionaries across all 'videos_info.json' files in '{parent_folder}': {total_dicts}")
    print(f"Total child folders in '{parent_folder}': {child_folders}")
    return total_dicts, child_folders


# Example usage
parent_folder = "data/raw/user_amthuc"  # Replace with your actual folder path
count_total_dicts_in_folder(parent_folder)
```
